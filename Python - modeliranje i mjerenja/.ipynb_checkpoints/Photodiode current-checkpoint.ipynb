```python
# SZA - 25 for summer, 70 for winter
# DU - Dobson units in atmosphere

### --- OVDJE SAM USPOREĐIVAO KOLIČINU OZONA KOJU DOBIJEM SA SMARTS-ovim INTENZITETIMA

def direct_irradiance(SZA, DU):
    
    ps = 820
    p0 = 1013.3
    
    ##------------- RELATIVNA OPTIČKA ZRAČNA MASA
    
    a1_o = 2.6845e2        # air-mass ozone
    a2_o = 0.5
    a3_o = 115.420
    a4_o = -3.2922
    air_mass_o = (np.cos(np.deg2rad(SZA)) + a1_o*SZA**a2_o * (a3_o-SZA)**a4_o)**-1

    a1_r = 4.5665e-1       # air-mass Rayleigh scattering
    a2_r = 0.07
    a3_r = 96.4836
    a4_r = -1.6970
    air_mass_R = (np.cos(np.deg2rad(SZA)) + a1_r*SZA**a2_r * (a3_r-SZA)**a4_r)**-1

    a1_ae = 3.1141e-2      # air-mass aerosol scattering
    a2_ae = 0.1
    a3_ae = 92.4710
    a4_ae = -1.3814
    air_mass_ae = (np.cos(np.deg2rad(SZA)) + a1_ae*SZA**a2_ae * (a3_ae-SZA)**a4_ae)**-1

    
    #-------------- Rayleighovo raspršenje
    
    aprox_ray = 1.787e10*wvlgth**-4.25*np.log(10)   # 1.787e10 x λ[nm]**-4.25
                                                    # a simple approximation for the base 10
    a1 = 117.2594*1e-12                             # SMARTS calculation
    a2 = -1.3215*1e-6
    a3 = 3.2073e-4
    a4 = -7.6842e-5*1e6
    exact_ray = (a1*(wvlgth)**4 + a2*(wvlgth)**2 + a3 + a4*(wvlgth)**-2)**-1
    ray = (ps/p0)*exact_ray
    
    #---------------- Apsorpcija na ozonu
    
    ozone = 2.687e16*cross_sec*DU
    
    #-------------- Raspršenje na aerosolima
    alfa1 = 0.932
    alfa2 = 1.441
    ang_beta = 0.0198
    beta = 2**(alfa2-alfa1)*ang_beta
    lam1 = 1000
    aerosol = beta*(wvlgth/lam1)**-alfa1
    
    scatt1 = (ray[80]-ray[60])*air_mass_R / (2.687e16*(cross_sec[80]-cross_sec[60])*air_mass_o)
    scatt2 = (aerosol[80]-aerosol[60])*air_mass_ae / (2.687e16*(cross_sec[80]-cross_sec[60])*air_mass_o)
    
    return [x*m.exp(-y*air_mass_o-z*air_mass_R-k*air_mass_ae) for x,y,z,k in zip(exterre,ozone,ray,aerosol)], scatt1, scatt2, air_mass_o

"""
ozone_surf_irr = [x*m.exp(-y*air_mass_o) 
                      for x,y in zip(exterre,ozone)]
"""

IRR300W, scatt1W, scatt2W, airmassW  = direct_irradiance(70, 300.0)
IRR300S, scatt1S, scatt2S, airmassS  = direct_irradiance(25, 300.0)

### -------- OZONE COLUMN FORMULA --------

print("'MLS', Z=25 : Rayleigh scattering and scattering on aerosols contributions are {} and {}".format(scatt1S, scatt2S))
ozon = (m.log(IRR300S[60]/IRR300S[80]) - m.log(exterre[60]/exterre[80])) / (2.687e16*(cross_sec[80]-cross_sec[60])*airmassS)
X = ozon - scatt1S - scatt2S
print("For a wavelength couple {} and {} nm: {} DU for code approximation\n".format(wvlgth[60], wvlgth[80], X))
###---------------------
print("'MLW', Z=70 : Rayleigh scattering and scattering on aerosols contributions are {} and {}".format(scatt1W, scatt2W))
ozon = (m.log(IRR300W[60]/IRR300W[80]) - m.log(exterre[60]/exterre[80])) / (2.687e16*(cross_sec[80]-cross_sec[60])*airmassW)
X = ozon - scatt1W - scatt2W
print("For a wavelength couple {} and {} nm: {} DU for code approximation".format(wvlgth[60], wvlgth[80], X))
###---------------------

###---------------------
print("\nTrebalo bi biti 270 DU :")
ozon1 = (m.log(DU270S[60]/DU270S[80]) - m.log(exterre[60]/exterre[80])) / (2.687e16*(cross_sec[80]-cross_sec[60])*airmassS)
ozon2 = (m.log(DU270W[60]/DU270W[80]) - m.log(exterre[60]/exterre[80])) / (2.687e16*(cross_sec[80]-cross_sec[60])*airmassW)
ozon3 = (m.log(DU270[60]/DU270[80]) - m.log(exterre[60]/exterre[80])) / (2.687e16*(cross_sec[80]-cross_sec[60])*airmassW)
X1 = ozon1 - scatt1S - scatt2S
X2 = ozon2 - scatt1W - scatt2W
X3 = ozon3 - scatt1W - scatt2W
print("'MLS', Z=25 : {} DU\n'MLW', Z=70 : {} DU\n'MLS', Z=70 : {} DU".format(X1, X2, X3))
###--------------------
print("\nTrebalo bi biti 300 DU :")
ozon1 = (m.log(DU300S[60]/DU300S[80]) - m.log(exterre[60]/exterre[80])) / (2.687e16*(cross_sec[80]-cross_sec[60])*airmassS)
ozon2 = (m.log(DU300W[60]/DU300W[80]) - m.log(exterre[60]/exterre[80])) / (2.687e16*(cross_sec[80]-cross_sec[60])*airmassW)
ozon3 = (m.log(DU300[60]/DU300[80]) - m.log(exterre[60]/exterre[80])) / (2.687e16*(cross_sec[80]-cross_sec[60])*airmassW)
X1 = ozon1 - scatt1S - scatt2S
X2 = ozon2 - scatt1W - scatt2W
X3 = ozon3 - scatt1W - scatt2W
print("'MLS', Z=25 : {} DU\n'MLW', Z=70 : {} DU\n'MLS', Z=70 : {} DU".format(X1, X2, X3))
###--------------------
print("\nTrebalo bi biti 330 DU :")
ozon1 = (m.log(DU330S[60]/DU330S[80]) - m.log(exterre[60]/exterre[80])) / (2.687e16*(cross_sec[80]-cross_sec[60])*airmassS)
ozon2 = (m.log(DU330W[60]/DU330W[80]) - m.log(exterre[60]/exterre[80])) / (2.687e16*(cross_sec[80]-cross_sec[60])*airmassW)
ozon3 = (m.log(DU330[60]/DU330[80]) - m.log(exterre[60]/exterre[80])) / (2.687e16*(cross_sec[80]-cross_sec[60])*airmassW)
X1 = ozon1 - scatt1S - scatt2S
X2 = ozon2 - scatt1W - scatt2W
X3 = ozon3 - scatt1W - scatt2W
print("'MLS', Z=25 : {} DU\n'MLW', Z=70 : {} DU\n'MLS', Z=70 : {} DU".format(X1, X2, X3))
###--------------------
```

```python
### --- KOLIKA STRUJA JE NA IZLAZU DETEKTORA
### --- i dalje koristim smartsove intenzitete sa krajnjim vrijednostima zenitnog kuta (zima i ljeto)

S = 0.06e-6

DU = [250, 260, 270, 280, 290, 300, 310, 320, 330, 340, 350]
IRR_DU_W = np.array([DU250W, DU260W, DU270W, DU280W, DU290W, DU300W, DU310W, DU320W, DU330W, DU340W, DU350W])
IRR_DU_S = np.array([DU250S, DU260S, DU270S, DU280S, DU290S, DU300S, DU310S, DU320S, DU330S, DU340S, DU350S])

I_SG01S_18_310nm_W=[]
I_SG01S_18_320nm_W=[]

I_SG01S_18_310nm_S=[]
I_SG01S_18_320nm_S=[]

for i in range(len(DU)):
    
    I = S * integrate.simps(IRR_DU_W[i] * detector1,dx=0.5) * 1e9
    I_SG01S_18_310nm_W.append(I)
    I = S * integrate.simps(IRR_DU_W[i] * detector2,dx=0.5) * 1e9
    I_SG01S_18_320nm_W.append(I)
    
    I = S * integrate.simps(IRR_DU_S[i] * detector1,dx=0.5) * 1e9
    I_SG01S_18_310nm_S.append(I)
    I = S * integrate.simps(IRR_DU_S[i] * detector2,dx=0.5) * 1e9
    I_SG01S_18_320nm_S.append(I)

ds = []
dw = []
for i in range(len(I_SG01S_18_320nm_W)):
    d = I_SG01S_18_320nm_W[i] / I_SG01S_18_310nm_W[i]
    dw.append(d)
    d = I_SG01S_18_320nm_S[i] / I_SG01S_18_310nm_S[i]
    ds.append(d)

dw = np.array(dw); ds = np.array(ds)
l1 = np.linspace(dw[5]-5*(dw[6]-dw[5]), dw[6]+4*(dw[6]-dw[5]), 11)
l2 = np.linspace(ds[5]-5*(ds[6]-ds[5]), ds[6]+4*(ds[6]-ds[5]), 11)
```

```python
### --- OVDJE SAM SIMULIRAO, U SMARTS-u, INTENZITETE ZA PROMJENU KUTA UZ ISTI OZON

SZA25 = ascii.read("SZA25.txt")['Direct_normal_irradiance'][:181]
SZA30 = ascii.read("SZA30.txt")['Direct_normal_irradiance'][:181]
SZA35 = ascii.read("SZA35.txt")['Direct_normal_irradiance'][:181]
SZA40 = ascii.read("SZA40.txt")['Direct_normal_irradiance'][:181]
SZA45 = ascii.read("SZA45.txt")['Direct_normal_irradiance'][:181]
SZA50 = ascii.read("SZA50.txt")['Direct_normal_irradiance'][:181]
SZA55 = ascii.read("SZA55.txt")['Direct_normal_irradiance'][:181]
SZA60 = ascii.read("SZA60.txt")['Direct_normal_irradiance'][:181]
SZA65 = ascii.read("SZA65.txt")['Direct_normal_irradiance'][:181]
SZA70 = ascii.read("SZA70.txt")['Direct_normal_irradiance'][:181]

I_310nm=[]
I_320nm=[]

SZA = np.array([25,30,35,40,45,50,55,60,65,70])
IRR_SZA = np.array([SZA25, SZA30, SZA35, SZA40, SZA45, SZA50, SZA55, SZA60, SZA65, SZA70])
```

```python
### --- ODAVDJE NADALJE SAM SIMULIRAM PROLAZAK SVJETLA KROZ ATMOSFERU UZ POMOĆ KODA KOJI SAM SLOŽIO ZA SEMINAR
### --- SVE OVE FORMULE SU IZ SMARTSOVOG PRIRUČNIKA

###   SIMULIRA ATMOSFERU ZA DAN ZENITNI KUT I KOLIČINU OZONA
###   VRAĆA IZNOSE OSVJETLJENJA NA KRAĆOJ I DUŽOJ VALNOJ DULJINI

def atmosphere(SZA, DU):
    ps = 1017
    p0 = 1013.3
    ##------------- RELATIVNA OPTIČKA ZRAČNA MASA
    a1_o = 2.6845e2        # air-mass ozone
    a2_o = 0.5
    a3_o = 115.420
    a4_o = -3.2922
    air_mass_o = (np.cos(np.deg2rad(SZA)) + a1_o*SZA**a2_o * (a3_o-SZA)**a4_o)**-1
    a1_r = 4.5665e-1       # air-mass Rayleigh scattering
    a2_r = 0.07
    a3_r = 96.4836
    a4_r = -1.6970
    air_mass_R = (np.cos(np.deg2rad(SZA)) + a1_r*SZA**a2_r * (a3_r-SZA)**a4_r)**-1
    a1_ae = 3.1141e-2      # air-mass aerosol scattering
    a2_ae = 0.1
    a3_ae = 92.4710
    a4_ae = -1.3814
    air_mass_ae = (np.cos(np.deg2rad(SZA)) + a1_ae*SZA**a2_ae * (a3_ae-SZA)**a4_ae)**-1
    #-------------- Rayleighovo raspršenje
    aprox_ray = 1.787e10*wvlgth**-4.25*np.log(10)   # 1.787e10 x λ[nm]**-4.25                                               
                                                    # a simple approximation for the base 10
    a1 = 117.2594*1e-12   # SMARTS calculation
    a2 = -1.3215*1e-6
    a3 = 3.2073e-4
    a4 = -7.6842e-5*1e6
    exact_ray = (a1*(wvlgth)**4 + a2*(wvlgth)**2 + a3 + a4*(wvlgth)**-2)**-1
    ray = (ps/p0)*exact_ray
    #---------------- Apsorpcija na ozonu
    ozone = 2.687e16*cross_sec*DU
    #-------------- Raspršenje na aerosolima
    alfa1 = 0.827
    alfa2 = 1.171
    tau_a5 = 0.3442 # SMARTS2 value
    # nisam sigurna a ovo, dobije se 0.153 dolje: ang_beta = 0.0198
    ang_beta = tau_a5/(2**alfa2)  # prema ovome je 0.153
    #print(ang_beta)  
    beta1 = 2**(alfa2-alfa1)*ang_beta
    lam1 = 1000 #nm
    aerosol = beta1*(wvlgth/lam1)**-alfa1
    
    SURF_IRR = [x*np.exp(-y*air_mass_o-z*air_mass_R-k*air_mass_ae) for x,y,z,k in zip(exterre,ozone,ray,aerosol)]
    return SURF_IRR


###   PRIMA SZA I PRIRODNI LOGARITAM OMJERA OSVJETLJENJA NA KRAĆOJ I DUŽOJ VALNOJ DULJINI
###   VRAĆA KOLIČINU OZONA U DOBSONOVIM JEDINICAMA

def total_ozone(SZA, IRR):
    ps = 820
    p0 = 1013.3
    ##------------- RELATIVNA OPTIČKA ZRAČNA MASA
    a1_o = 2.6845e2        # air-mass ozone
    a2_o = 0.5
    a3_o = 115.420
    a4_o = -3.2922
    airmass_o = (np.cos(np.deg2rad(SZA)) + a1_o*SZA**a2_o * (a3_o-SZA)**a4_o)**-1
    a1_r = 4.5665e-1       # air-mass Rayleigh scattering
    a2_r = 0.07
    a3_r = 96.4836
    a4_r = -1.6970
    airmass_R = (np.cos(np.deg2rad(SZA)) + a1_r*SZA**a2_r * (a3_r-SZA)**a4_r)**-1
    a1_ae = 3.1141e-2      # air-mass aerosol scattering
    a2_ae = 0.1
    a3_ae = 92.4710
    a4_ae = -1.3814
    airmass_ae = (np.cos(np.deg2rad(SZA)) + a1_ae*SZA**a2_ae * (a3_ae-SZA)**a4_ae)**-1
    #-------------- Rayleighovo raspršenje
    aproxray = 1.787e10*wvlgth**-4.25*np.log(10)   # 1.787e10 x λ[nm]**-4.25
                                                   # a simple approximation for the base 10
    a1 = 117.2594*1e-12    # SMARTS calculation
    a2 = -1.3215*1e-6
    a3 = 3.2073e-4
    a4 = -7.6842e-5*1e6
    smartsray = (a1*(wvlgth)**4 + a2*(wvlgth)**2 + a3 + a4*(wvlgth)**-2)**-1
    ray = (ps/p0)*smartsray
    #-------------- Raspršenje na aerosolima
    alfa1 = 0.827
    alfa2 = 1.171
    ang_beta = 0.0198
    beta = 2**(alfa2-alfa1)*ang_beta
    lam1 = 1000
    aerosol = beta*(wvlgth/lam1)**-alfa1
    
    N = np.log(IRR[60]/IRR[80])
    
    scatt1 = (ray[80]-ray[60])*airmass_R / (2.687e16*(cross_sec[80]-cross_sec[60])*airmass_o)
    scatt2 = (aerosol[80]-aerosol[60])*airmass_ae / (2.687e16*(cross_sec[80]-cross_sec[60])*airmass_o)
    ozon = (N - np.log(exterre[60]/exterre[80])) / (2.687e16*(cross_sec[80]-cross_sec[60])*airmass_o)
    X = ozon - scatt1 - scatt2
    print("Za SZA = {} stupnjeva, količina ozona je {} DU".format(SZA,X))  
    return

### --- trebalo bi biti 300 DU, uzeti su SMARTS intenziteti za razne SZA iz gornjeg bloka

for i in range(len(SZA)):  
    total_ozone(SZA[i],IRR_SZA[i])
```

```python
def atmosphere(SZA, DU, duNO2):
    ps = 29.89 * 33.86389 # inHg to hPa
    p0 = 1013.3
    ##------------- RELATIVNA OPTIČKA ZRAČNA MASA
    a1_o = 2.6845e2        # air-mass ozone
    a2_o = 0.5
    a3_o = 115.420
    a4_o = -3.2922
    air_mass_o = (np.cos(np.deg2rad(SZA)) + a1_o*SZA**a2_o * (a3_o-SZA)**a4_o)**-1
    a1_r = 4.5665e-1       # air-mass Rayleigh scattering
    a2_r = 0.07
    a3_r = 96.4836
    a4_r = -1.6970
    air_mass_R = (np.cos(np.deg2rad(SZA)) + a1_r*SZA**a2_r * (a3_r-SZA)**a4_r)**-1
    a1_ae = 3.1141e-2      # air-mass aerosol scattering
    a2_ae = 0.1
    a3_ae = 92.4710
    a4_ae = -1.3814
    air_mass_ae = (np.cos(np.deg2rad(SZA)) + a1_ae*SZA**a2_ae * (a3_ae-SZA)**a4_ae)**-1
    #-------------- Rayleighovo raspršenje
    aprox_ray = 1.787e10*wvlgth**-4.25*np.log(10)   # 1.787e10 x λ[nm]**-4.25                                               
                                                    # a simple approximation for the base 10
    a1 = 117.2594*1e-12   # SMARTS calculation
    a2 = -1.3215*1e-6
    a3 = 3.2073e-4
    a4 = -7.6842e-5*1e6
    exact_ray = (a1*(wvlgth)**4 + a2*(wvlgth)**2 + a3 + a4*(wvlgth)**-2)**-1
    ray = (ps/p0)*exact_ray
    #---------------- Apsorpcija na ozonu
    ozone = 2.687e16*cross_sec*DU
    #---------------- Apsorpcija na NO2
    NO2 = 2.687e16*crossNO2*duNO2
    #-------------- Raspršenje na aerosolima
    alfa1_0 = 0.822 # urban 0% humidity
    alfa1_50 = 0.827 # urban 50% humidity
    alfa2_0 = 1.167 # urban 0% humidity
    alfa2_50 = 1.171 # urban 50% humidity
    tau_a5 = 0.3442 # SMARTS2 value
    # nisam sigurna a ovo, dobije se 0.153 dolje: ang_beta = 0.0198
    ang_beta = tau_a5/(2**alfa2_50)  # prema ovome je 0.153
    #print(ang_beta)  
    beta1 = 2**(alfa2_50-alfa1_50)*ang_beta
    lam1 = 1000 #nm
    aerosol = beta1*(wvlgth/lam1)**-alfa1_50
    
    SURF_IRR = [x*np.exp(-y*air_mass_o-z*air_mass_R-k*air_mass_ae-l) for x,y,z,k,l in zip(exterre,ozone,ray,aerosol,NO2)]
    return SURF_IRR

# Ovo su snage, bez da je uzeta u obzir aktivna površina, cilj je samo dobiti omjer
def detector310_power(SZA, DU,duNO2):
    return integrate.simpson(detector1*atmosphere(SZA, DU,duNO2),x=wvlgth)
def detector320_power(SZA, DU,duNO2):
    return integrate.simpson(detector2*atmosphere(SZA, DU,duNO2),x=wvlgth)
def ratio_310_over_320(SZA, DU,duNO2):
    return detector310_power(SZA, DU,duNO2)/detector320_power(SZA, DU,duNO2)
```
