#### Absorption coefficient and column density

Absorption coefficient $\alpha_\nu$ (in cm) is defined from the empirical law as the constant of proportionality between the amount of radiation lost and incident radiation:
$$dI_\nu = - \alpha_\nu\,ds\, I_\nu$$

Microscopic explanation:

If we have a number density $n$ of radiation absorbers which are randomly distributed, then the total number of them in a volume $dA\,ds$ is $n\,dA\,ds$. If each absorber has an effective absorbing area called the cross-section $\sigma_nu$, their total absorbing area in the volume is $\sigma_nu \,N\,dA\,ds$.

Energy absorbed by the beam is $I_\nu (\sigma_\nu \,n\,dA\,ds) dt\,d\nu\,d\Omega$, which is equal to the loss of the beam energy: $- dI_\nu \,dA dt\,d\nu\,d\Omega $.

$$I_\nu  \sigma_\nu \,n\,ds = - dI_\nu $$

This means that $\alpha_\nu = n \,\sigma_\nu$.

So, the radiation passing from $s_0$ to $s$:

$$I_\nu(s) = I_\nu(s_0) \exp(-\int_{s_0}^s \alpha_\nu(s')ds')$$

We can define the optical depth $\tau$ as

$$\tau_\nu(s) = \int_{s_0}^s \alpha_\nu(s')ds'$$

and loss equation is more simply written as

$$\frac{dI_\nu}{d\tau_\nu} = - I_\nu$$ 

and $d\tau_\nu = \alpha_\nu ds$

#### Dobson units

Take a cylinder with height $h$ and surface area of $S$, therefore the total number of air molecules in the cylinder would be:
$$N = nhS$$

So, the number of molecules per unit area of the base of the cylinder is $N/S = n\,h $.


1 DU is defined as the thickness of the layer in 10µm which would be formed by the total column amount at STP.  
$$1{\rm DU} = N/S = n\cdot 10µm = 2.6516e20 \,\rm{m}^{-2}$$

Therefore DU counts the *number of molecules per meter squared* in units/multiples of 2.6516e20.
