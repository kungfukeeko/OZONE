```python
import numpy as np
import scipy
import matplotlib.pyplot as plt
from astropy.io import ascii

i,azimuth,elevation,ratio,v1,v2 = np.loadtxt('ozone_3_1.csv', delimiter=",", dtype = 'float', unpack=True)
window = int(len(ratio))
meanRatio = scipy.signal.savgol_filter(ratio, window,3)
```

```python
time = np.arange(0,len(elevation),1)
fig, ax = plt.subplots(figsize=(16,5))
ax.plot(time, elevation, linestyle='-')
ax.set(xlabel='Time', ylabel='elevation angle',title='')
ax.grid()
plt.show()
plt.close(fig)
```

```python
fig, ax = plt.subplots(2,1,figsize=(16,12))
ax[0].plot(i, v1, linestyle='-',label='Napon na izlazu prvog pojačala (310nm)')
ax[0].plot(i, v2, linestyle='-',label='Napon na izlazu drugog pojačala (320nm)')
ax[0].set(xlabel='Time', ylabel='output voltage',title='Naponi na izlazu')
ax[0].grid()
ax[0].legend(loc='best')
ax[1].plot(i/60, ratio, linestyle='-')
ax[1].plot(i/60, meanRatio, linestyle='-')
ax[1].set(xlabel='Time', ylabel='intensity ratio (310/320)',title='Omjer intenziteta (310/320)',ylim=(0,1))
ax[1].grid()
plt.show()
plt.close(fig)
```
