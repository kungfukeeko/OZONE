```python
import numpy as np
import matplotlib.pyplot as plt
from astropy.io import ascii
```

### PuTTY log 2023.06.24 15:12:37 - ovo je vrijeme početka mjerenja, a na X-osi su minute trajanja
#### Prema Suncu, oblak prolazi

```python
i,v1,v2,ratio = np.loadtxt('ozone_no_filter_diffuse.txt', delimiter=",", dtype = 'float', unpack=True)
fig, ax = plt.subplots(2,1,figsize=(16,12))
ax[0].plot(i/60, v1, linestyle='-',label='Napon na izlazu kanala 1)')
ax[0].plot(i/60, v2, linestyle='-',label='Napon na izlazu kanala 2)')
ax[0].set(xlabel='Vrijeme [min]', ylabel='[V]',title='Naponi na izlazu')
ax[0].grid()
ax[0].legend(loc='best')
ax[1].plot(i/60, ratio, linestyle='-')
ax[1].set(xlabel='Vrijeme [min]', ylabel='Omjer intenziteta (kanal 1 / kanal 2)',title='Omjer intenziteta (kana1/kanal2)',ylim=(0.5,1.5))
ax[1].grid()
plt.show()
plt.close(fig)
np.mean(ratio)
```

### 2023.06.24 18:42 Napravljeno usrednjavanje na Arduinu
### Usrednjavanje 20 sampleova svake dvije sekunde
#### Oblačno, Sunce je zašlo iza zgrade

```python
i, meanV1, meanV2, meanRatio, stdV1, stdV2, stdRatio = np.loadtxt('ozone_no_filter_diffuse_mean_std.csv', delimiter=",", dtype = 'float', unpack=True)
fig, ax = plt.subplots(2,1,figsize=(16,12))
ax[0].plot(i/60, meanV1, linestyle='-',label='Napon na izlazu prvog pojačala (310nm)')
ax[0].plot(i/60, meanV2, linestyle='-',label='Napon na izlazu drugog pojačala (320nm)')
ax[0].set(xlabel='Time', ylabel='output voltage',title='Naponi na izlazu')
ax[0].grid()
ax[0].legend(loc='best')
ax[1].plot(i/60, meanRatio, linestyle='-')
ax[1].errorbar(i/60, meanRatio, stdRatio)
ax[1].set(xlabel='Time', ylabel='intensity ratio (310/320)',title='Omjer intenziteta (310/320)',ylim=(0.5,1.5))
ax[1].grid()
plt.show()
plt.close(fig)

np.mean(meanRatio)
```
